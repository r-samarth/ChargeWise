# 01 — Data Collection & Source Provenance (ChargeWise Karnataka)

**Objective:** Load and inspect all raw datasets supporting ChargeWise's Karnataka charging intelligence and DAV academic analysis.

### Documented Sources & Licenses
1. **OpenStreetMap Overpass API (`data/raw/osm_karnataka_chargers.json`)** — Retrieved `2026-11-25` under **ODbL 1.0**. Raw bounding-box query `(11.5, 74.0, 18.5, 78.6)`.
2. **PostgreSQL/PostGIS Curated + OSM Export (`data/processed/karnataka_stations_raw_export.csv`)** — Exported from ChargeWise PostgreSQL database combining verified highway disclosures (Tata Power, Zeon, Statiq) with strict Karnataka OSM records.
3. **VAHAN / BEE India EV Market Dataset (`data/raw/india_ev_market/`)** — Monthly state-level & maker-level EV sales (`Apr 2021 – Mar 2024`) plus BEE state-wise public charging station counts (`dim_public_charging_stations_by_state.csv`).
4. **Excluded Source Note:** NITI Aayog e-Amrit data was evaluated and intentionally excluded due to website terms requiring prior written permission.

```python
import json
from pathlib import Path
import pandas as pd

REPO = Path("..").resolve()
raw_osm_path = REPO / "data" / "raw" / "osm_karnataka_chargers.json"
db_stations_path = REPO / "data" / "processed" / "karnataka_stations_raw_export.csv"
ev_dir = REPO / "data" / "raw" / "india_ev_market"

osm_raw = json.loads(raw_osm_path.read_text(encoding="utf-8"))
stations_df = pd.read_csv(db_stations_path)
sales_state_df = pd.read_csv(ev_dir / "electric_vehicle_sales_by_state.csv")
sales_makers_df = pd.read_csv(ev_dir / "electric_vehicle_sales_by_makers.csv")
bee_stations_df = pd.read_csv(ev_dir / "dim_public_charging_stations_by_state.csv")

print(f"Raw OSM Overpass bbox elements : {len(osm_raw.get('elements', []))}")
print(f"ChargeWise Karnataka stations  : {stations_df.shape}")
print(f"VAHAN sales by state           : {sales_state_df.shape}")
print(f"VAHAN sales by maker           : {sales_makers_df.shape}")
print(f"BEE public stations by state   : {bee_stations_df.shape}")
```
